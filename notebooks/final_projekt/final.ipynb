## Estimation of Obesity Levels Based On Eating Habits and Physical Condition

მახასიათებლების აღწერა:
| Column Name                           | Type                                                                                 | Description                                                                                         |
| ------------------------------------- | ------------------------------------------------------------------------------------ | --------------------------------------------------------------------------------------------------- |
| Gender                            | Categorical (“Male” / “Female”)                                                      | Individual’s gender                                                                                |
| Age                               | Continuous (years)                                                                   | Person’s age in whole years                                                                        |
| Height                            | Continuous (centimeters)                                                             | Individual’s height measured in centimeters                                                        |
| Weight                            | Continuous (kilograms)                                                               | Person’s weight measured in kilograms                                                              |
| family\_history\_with\_overweight | Binary (“yes” / “no”)                                                                | Indicates whether the individual has or had a family member who is/was overweight                  |
| FAVC                              | Binary (“yes” / “no”)                                                                | Frequent consumption of high‐caloric food (ie, often eats calorie‐dense foods)                   |
| FCVC                              | Integer (1–3)                                                                        | Frequency of vegetable consumption in meals (1 = never, 2 = sometimes, 3 = always)                 |
| NCP                               | Integer (1–4)                                                                        | Number of main meals consumed per day (ranges from 1 to 4)                                         |
| CAEC                              | Categorical (“no,” “Sometimes,” “Frequently,” “Always”)                              | Consumption of food between meals (snacking behavior)                                              |
| SMOKE                             | Binary (“yes” / “no”)                                                                | Whether the individual smokes tobacco                                                              |
| CH2O                              | Integer (1–3)                                                                        | Daily water consumption: 1 = < 1 L, 2 = 1–2 L, 3 = > 2 L per day                                   |
| SCC                               | Binary (“yes” / “no”)                                                                | Whether the person monitors their caloric intake (eg, tracks calories)                           |
| FAF                               | Integer (0–3)                                                                        | Frequency of physical activity per week (0 = never, 1–3 = increasing levels of activity frequency) |
| TUE                               | Integer (0–2)                                                                        | Time spent using technology devices daily: 0 = no use/sedentary, 1 = moderate use, 2 = high use    |
| CALC                              | Categorical (“no,” “Sometimes,” “Frequently,” “Always”)                              | Consumption of alcohol (frequency of drinking alcoholic beverages)                                 |
| MTRANS                            | Categorical (“Automobile,” “Bike,” “Motorbike,” “Public\_Transportation,” “Walking”) | Primary transportation mode used by the individual                                                 |
| NObesity                          | Categorical (seven levels)                                                           | Target variable indicating obesity category                                                      |                                                                                     |                                                                                                     |

### 3. Classification Task
- "NObesity" არის სამიზნე სვეტი (y)
- დაყავით მონაცემები Train/Validation/Test (მაგ.: 60% / 20% / 20%) სიმრავლეებად
- გამოიყენეთ რიცხვითი მონაცემებისათვის StandardScaler და კატეგორიული მახასიათებლებსთვის One-Hot Encoding
- გამოიყენეთ `LogisticRegression` L2-Regularization-ით (`LogisticRegression(max_iter=1000, penalty='l2')`) როგორც საბაზისო მოდელი
  - გაწვრთენით `LogisticRegression` მოდელი Train მონაცემებზე
  - შეაფასე Validation მონაცემებზე: Accuracy, Precision, Recall, F1-score, Confusion Matrix
- მახასიათებლების შერჩევა: გამოიყენე `RFE` (Logistic Regression) 10 ყველაზე მნიშვნელოვანი მახასიათებლის შესარჩევად
- შერჩეულ მახასიათებლებზე გაწვრთენით რამდენიმე მოდელი
 * `Decision Tree`, `Random Forest` ან `Gradient Boosting`
 * `SVM` (RBF)
 * `k-Nearest Neighbors`
- Validation მონაცემებზე დაყრდნობით შეარჩიეთ მოდელი, რომელსაც აქვს საუკეთესო პერფორმანსი
- გამოიყენეთ ჰიპერპარამეტრების ოპტიმიზაცია (GridSearchCV)
- საუკეთესო მოდელის ევალუაცია Test სიმრავლეზე: Accuracy, Precision, Recall, F1-score, Confusion Matrix

### 4. Regression Task
- დაამატეთ სვეტი `df_obesity["BMI"] = df_obesity["Weight"] / (df_obesity["Height"] ** 2)`
- "Weight", "Height", "NObesity" - ეს სვეტები არ გაითვალისწინოთ 
- "BMI" არის სამიზნე სვეტი (y)
- დაყავით მონაცემები Train/Validation/Test (მაგ.: 60% / 20% / 20%) სიმრავლეებად
- გამოიყენეთ რიცხვითი მონაცემებისათვის StandardScaler 
- გამოიყენეთ `LinearRegression` როგორც საბაზისო მოდელი

- Validation მონაცემებზე დაყრდნობით შეარჩიეთ მოდელი, რომელსაც აქვს საუკეთესო პერფორმანსი
