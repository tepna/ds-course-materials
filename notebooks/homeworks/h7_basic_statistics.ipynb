### 1. DataFrame-ის ფაილიდან წაკითხვა, მონაცემების ანალიზი

ამ დავალებისთვის გამოვიყენებთ Amazon-ის გაყიდვების მონაცემებს (Fake Data).

  - გამოიტანეთ მონაცემების overview `head()` მეთოდის გამოყენებით.
    
- **სტატისტიკური ანალიზი**
  - გამოთვალეთ და დაბეჭდეთ საშუალო შესყიდვის ფასი (Purchase Price)
  - გამოთვალეთ და დაბეჭდეთ შესყიდვის ფასის მედიანა 
  - დაბეჭდეთ ყველაზე მაღალი და ყველაზე დაბალი შესყიდვების ფასის მნიშვნელობა
  - გამოთვალეთ და დაბეჭდეთ შესყიდვის ფასის სტანდარტული გადახრა
  - გამოთვალეთ და დაბეჭდეთ შესყიდვის ფასის 25%, 75% კვანტილები
  - მიიღეთ შესყიდვის ფასის მონაცემთა აღწერა describe() მეთოდის საშუალებით
  - ჰისტოგრამის საშუალებით (Purchase Price) მონაცემების ვიზუალიზაცია, რათა გამოვიკვლიოთ მონაცემთა განაწილება
  - გამოიყენეთ ბოქს-პლოტი (Purchase Price) მონაცემების ვიზუალიზაციისათვის, რათა გამოვიკვლიოთ მონაცემთა განაწილება.
  - Outliers აღმოჩენა კვანტილების გამოყენებით (IQR)
