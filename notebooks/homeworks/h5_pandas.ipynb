### 1. Pandas Series-ის შექმნა და ოპერაციების განხორციელება
- **Series-ის შექმნა:**  
  შექმენით Series მონაცემების სიიდან და მიანიჭე შესაბამისი ინდექსი, მაგალითად
  - მონაცემები: `["12.5", "18.2", "21.3", "16.8", "19.5"]`
  - ინდექსი: `["D1", "D2", "D3", "D4", "D5"]`  
  შემდეგ, კონვერტაციის მეშვეობით შეცვალეთ მონაცემების ტიპი და მიანიჭეთ სვეტს`float` ტიპი.

- **სტატისტიკური ოპერაციები:**  
  - გამოიანგარიშეთ Series-ის საშუალო, მაქსიმალური და მინიმალური მნიშვნელობები. გამოიყენეთ ფუნქციები `mean()`, `max()`, `min()`.

- **ელემენტებზე ფუნქციების შეტანა:**  
  - შექმენით ახალი Series, სადაც თითოეული ელემენტი ითვლება ორიგინალური მნიშვნელობის ორზე გამრავლებით, გამოიყენეთ ვექტორული ოპერაციები.

- **ფილტრაცია და შერჩევა:**  
  - დაწერეთ ფუნქცია, რომელიც იღებს Series-ს და აბრუნებს ახალ Series-ს მხოლოდ იმ ელემენტებით, რომლებიც აღემატება მოცემულ რიცხვს (მაგ., tolerance = 18).

````python
# Your code here
import pandas as pd

data = ["12.5", "18.2", "21.3", "16.8", "19.5"]
index = ["D1", "D2", "D3", "D4", "D5"]

examp = pd.Series(data=data, index=index)
examp = examp.astype("float")

examp_mean = examp.mean()
examp_min = examp.min()
examp_max = examp.max()

examp_multiplied_two = examp *2

def filt(data,tolerance):
    return data[data > tolerance]
````

### 2. DataFrame-ის შექმნა და ოპერაციების განხორციელება

- **DataFrame-ის შექმნა:**  
  - შექმენი DataFrame მოცემული dictionary-დან:
```
data = {
    "მომხმარებელი": [1, 2, 3, 4, 5],
    "ანაზღაურება": [3000, 2500, 2000, 4500, 4000],
    "საკრედიტო ისტორია": [5, 2, 1, 4, 3],
    "მიმდინარე დავალიანება": [1000, 900, 0, 2000, 1500],
    "გადახდის სიზუსტე": [95, 90, 75, 95, 90]
}
```
- **საბაზისო მიმოხილვა:**  
  - გამოიტანეთ მონაცემების overview `head()` და `tail()` მეთოდების გამოყენებით.
  - მიიღეთ მონაცემთა აღწერა `describe()` და `info()` მეთოდების დახმარებით.

- **ფილტრაცია და სვეტების დამატება:**  
  - **ფილტრაცია:**
    - აირჩიეთ მხოლოდ რიგები 1:3
    - აირჩიეთ მხოლოდ ის ჩანაწერები, სადაც `ანაზღაურება` მეტია 2500-ზე.
  - **დამატება:** დაამატეთ ახალი სვეტი "საკრედიტო ქულა":
    - $\text{საკრედიტო ქულა} = 0.055 \times \text{ანაზღაურება} + 6.46 \times \text{გადახდის სიზუსტე}$

- **CSV ფაილთან მუშაობა:**  
  - შეინახეთ მიღებული DataFrame `to_csv()` მეთოდის გამოყენებით ახალ CSV ფაილში.

````python
# Your code here
data = {
    "მომხმარებელი": [1, 2, 3, 4, 5],
    "ანაზღაურება": [3000, 2500, 2000, 4500, 4000],
    "საკრედიტო ისტორია": [5, 2, 1, 4, 3],
    "მიმდინარე დავალიანება": [1000, 900, 0, 2000, 1500],
    "გადახდის სიზუსტე": [95, 90, 75, 95, 90]
}

data = pd.DataFrame(data)
data.head()
data.tail()
data.describe()
data.info()

data[1:3]
data[data["ანაზღაურება"] > 2500]

data["საკრედიტო ქულა"] = data["ანაზღაურება"] * 0.055 + 6.46* data["გადახდის სიზუსტე"]
data.to_csv("deskto/test.csv")
````
