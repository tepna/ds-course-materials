## **დავალებები: Data Exploration**

### ამ დავალებისთვის გამოვიყენებთ Adult (Census Income) მონაცემებს:

| Column            | Description                                 |
|-------------------|--------------------------------------------------------------------------------------------------------------|
| **age**           | Age of the individual in years.                                                                             |
| **workclass**     | Type of employment (e.g., Private, Self-emp-not-inc, Federal-gov, etc.).                                     |
| **fnlwgt**        | Final weight representing the number of people the census entry stands for.                                  |
| **education**     | Highest level of education attained (e.g., Bachelors, HS-grad, Masters, etc.).                               |
| **education.num** | Numerical representation of education level (often corresponding to years of education).                       |
| **marital.status**| Marital status of the individual (e.g., Married-civ-spouse, Divorced, Never-married, etc.).                    |
| **occupation**    | Type of occupation or job (e.g., Tech-support, Craft-repair, Sales, etc.).                                   |
| **relationship**  | Relationship to the household head (e.g., Husband, Wife, Own-child, Not-in-family, etc.).                      |
| **race**          | Racial background (e.g., White, Black, Asian-Pac-Islander, Amer-Indian-Eskimo, Other).                         |
| **sex**           | Gender of the individual (Male or Female).                                                                   |
| **capital.gain**  | Income received from investment sources, excluding wages (if any).                                             |
| **capital.loss**  | Losses incurred from investment sources.                                                                      |
| **hours.per.week**| Number of hours worked per week.                                                                              |
| **native.country**| Country of origin or where the individual was born.                                                          |
| **income**        | Income category of the individual (e.g., `<=50K` or `>50K`), indicating if the annual income exceeds a threshold.|


### Adult data მონაცემების ანალიზი
- **მონაცემების ფაილიდან წაკითხვა**
  - წაიკითხეთ მონაცემები adult.csv ფაილიდან (ფაილი მოთავსებულია: "../../datasets/adult_data/adult.csv")..
  - გამოიყენეთ მეთოდები, როგორიცაა `.head()`, `.info()` და `.describe()` მონაცემთა ფორმატსა და შიგთავსზე საწყისი შთაბეჭდილების შესაქმნელად.

  - დაბეჭდეთ იმ ინდივიდის მონაცემები, ვინც ასაკით არის ყველაზე ახალგაზრდა (age).

- **მონაცემების ანალიზი**
    - რიცხვითი ცვლადები: შექმენით ჰისტოგრამები, ბოქსპლოტები  რიცხვითი თვისებებისთვის, რა დისტრიბუცია აქვს თითოეული რიცხვით ცვლადს (მაგ., ასაკი, სამუშაო საათები) და ჩართეთ outlier-ების აღმოჩენა.
    - კატეგორიული ცვლადები: შექმენით ბარ-გრაფიკები, რათა ნახოთ კატეგორიული მახასიათებლების განაწილება, გამოიტანეთ თითოეული კატეგორიის (მაგ., workclass, education, occupation, income) განაწილების მაჩვენებლები და პროცენტული ოდენობა.
    - განხილეთ, რომელ ცვლადებს აქვთ ერთმანეთთან ძლიერი კორელაცია.
    - კორელაციური ანალიზი: გამოიკვლიეთ, არსებობს თუ არა რთული ურთიერთობები ციფრული ცვლადებს შორის (მაგ., ასაკი და სამუშაო საათები).
    - კატეგორიული და ციფრული ცვლადების ურთიერთობა: შეამოწმეთ, როგორ ითვლება შემოსავალი (income) სხვადასხვა კატეგორიებში, მაგალითად, race, gender, ან education-ის მიხედვით.
    - დააკვირდით, თუ როგორაა განაწილებული შემოსავალი (income) კატეგორიებით (მაგ., <=50K და >50K).
    - გამოიყენეთ groupby ოპერაციები, რათა გაარკვიოთ, როგორ განსხვავდება შემოსავალი სხვადასხვა ჯგუფებში (მაგ., სქესი, დასახლების ადგილი, საგანმანათლებლო დონე).
    - შექმენით ვიზუალიზაციები ისე, რომ თქვენი ანალიზის შედეგები მკაფიო და გასაგები იყოს. აღწერეთ გრაფიკები და განხილეთ, თუ რა დასკვნები შეგიძლიათ მიიღოთ.

```python
# - **მონაცემების ფაილიდან წაკითხვა**
#   - წაიკითხეთ მონაცემები adult.csv ფაილიდან (ფაილი მოთავსებულია: "../../datasets/adult_data/adult.csv")..
#   - გამოიყენეთ მეთოდები, როგორიცაა `.head()`, `.info()` და `.describe()` მონაცემთა ფორმატსა და შიგთავსზე საწყისი შთაბეჭდილების შესაქმნელად.
#   - დააკვირდით, რომელი სვეტი წარმოადგენს სამიზნე ცვლადს (რომელი სვეტის პროგნოზს გააკეთებდით)
import pandas as pd

data = pd.read_csv(r"C:\Users\amiri\OneDrive\Desktop\Projcets\Commshcool_DS\ds-course-materials\datasets\adult_dataset\adult.csv")
data.head()
data.describe()
data.info()
```

```python
# - **მონაცემთა გაწმენდა**
#   - შეამოწმეთ რომელ სვეტებშია None მნიშვნელობები
#   - გადაწყვიტეთ, როგორ მოიქცევით ამ None მნიშვნელობებთან მიმართებაში - იქნება ეს მნიშვნელობების შევსება, წაშლა თუ სხვა მეთოდის გამოყენება.
#   - გამოიკვლიეთ, არსებობს თუ არა სიმბოლოების (მაგ., “?”) სახით მოცემული არასწორი მონაცემები.
#   - დარწმუნდით, რომ ყველა სვეტი შეიცავს შესაბამის ტიპს (მაგ. რიცხვითი და კატეგორიული).

data.isna().any()
result = []
for col in data.columns:
    if (data[col] == "?").any():
        result.append(col)
```

```python
# - **მონაცემების გააზრება**
#   - დაადგინეთ მაქსიმალური კაპიტალური მოგება (capital.gain).
#   - დაადგინეთ მინიმალური capital.loss-ის მნიშვნელობა, დააკვირდით შედეგს.
#   - იპოვეთ ინდივიდი, რომელსაც ყველაზე მაღალი საგანმანათლებლო დონე აქვს (education.num) და დაბეჭდეთ მისი სამუშაო სფერო (occupation).
#   -  დაბეჭდეთ იმ ინდივიდის income-ს კატეგორია, რომელსაც მაქსიმალური hours.per.week-ს მნიშვნელობა აქვს.
#   - დაბეჭდეთ იმ ინდივიდის მონაცემები, ვინც ასაკით არის ყველაზე ახალგაზრდა (age).
#   - გამოთვალეთ თითოეული workclass-ის მიხედვით საშუალო (mean) capital.gain:
#    დაჯგუფებით (groupby) გაანგარიშეთ, რომელი სამუშაო კლასი იღებს მეტ კაპიტალურ მოგებას.
#   - რამდენი უნიკალური occupation არსებობს?
data["capital.gain"].max()
data["capital.gain"].min()
data[data["education.num"] == data["education.num"].max() ]["occupation"].unique()
data[data["hours.per.week"] == data["hours.per.week"].max()]["income"].unique()
data[data["age"] == data["age"].min()]
data.groupby("workclass").agg(
    MEAN_CAPITAL_GAIN = ("capital.gain","mean")
).reset_index()
data.groupby("workclass").agg(
    capital_gain_total = ("capital.gain","sum")
).reset_index()
data["occupation"].nunique()
```
