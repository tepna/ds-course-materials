````python
# Your code here

x,y = 4.5,7.8
multiply = x*y
x_sqrd = x**2
adding = x+y
subtract = x-y


if multiply > x_sqrd:
    print(f"Multiplication is greater than Squaring")
else:
     print(f"Squaring is greater than Multiplication")


if adding > subtract:
    print(f"Adding is greater than Subtracting")
else:
     print(f"Subtracting is greater than Adding")
````

## 7. ფუნქციები

7.1. შექმენით ფუნქცია `factorial(n)`, რომელიც იღებს რიცხვს და აბრუნებს მის ფაქტორიალს.

````python
# Your code here
x = 5
def factiorial(n):
    result = 1
    for i in range(1,n+1):
        result = result * i
    return result
  

factiorial(x)
````

7.2. დაწერეთ ფუნქცია, რომელიც იღებს სიას და აბრუნებს მასში მეორე ყველაზე პატარა რიცხვს.

````python
# Your code here

list_test = [1,2,4,8,4,324,77,2]

def idonno(x: list) -> int:
    x.sort()
    return x[1]

idonno(list_test)
````
