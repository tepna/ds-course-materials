## **დავალებები: Clustering¶**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
```

ამ დავალებისთვის გამოვიყენებთ Mall Customer მონაცემებს.

### Segment customers based on behavior

```python
# მონაცემების ჩატვირთვა
df_customers = pd.read_csv("../../datasets/mall_data/mall_customers.csv")
df_customers.head()
```

```python
# გამოვყოთ მახასიათებლები - features for clustering
X = df_customers[["Annual Income (k$)", "Spending Score (1-100)"]]
```

### KMeans
- გამოიყენეთ StandardScaler X სიმრავლის სკალირებისთვის
- სკალირებულ მონაცემებზე გაწვრთენით kmeans ალგორითმი (n_clusters=5)
- გამოთვალეთ თითოეული კლასტერისთვის საშუალო ასაკი, Annual Income და Spending Score
- შექმენით scatter გრაფიკი Annual Income vs. Spending Score (1-100)
- ასახეთ გრაფიკზე ალგორითმის საზღვრები და თითოეული კლასტერის ცენტი
- ააგეთ elbow rule, silhouette score გრაფიკები, იყო თუ არა n_clusters=5 სწორი გადაწყვეტილება?

### Cluster blobs

```python
X, y = make_circles(n_samples=1000, factor=0.5, noise=0.05)

plt.scatter(X[:, 0], X[:, 1], c='gray', edgecolor='k')
plt.title("Original Data (make_blobs)")
plt.show()
```

### DBSCAN
- გამოიყენეთ StandardScaler X სიმრავლის სკალირებისთვის
- სკალირებულ მონაცემებზე გაწვრთენით DBSCAN ალგორითმი, სწორად შეარჩიეთ eps, min_samples პარამეტრები
- ააგეთ გრაფიკი და განსხვავებული ფერით გამოყავით თითოეული კლასტერი
- ასახეთ გრაფიკზე ალგორითმის საზღვრები
- რა იქნებოდა აღებული min_samples-თვის ოპტიმალური eps პარამეტრი?

### OPTICS
- გამოიყენეთ StandardScaler X სიმრავლის სკალირებისთვის
- იმავე მონაცემებზე გაწვრთენით OPTICS ალგორითმი, გამოიყენეთ DBSCAN-თვის შერჩეული min_samples პარამეტრი
- ააგეთ Reachability Plot და დაადგინეთ რამდენ განსხვავებულ კლასტერს ხედავთ სიმჭიდროვის მიხედვით
- აარჩიეთ გრაფიკზე დაყრდნობით xi პარამეტრი
- ააგეთ გრაფიკი და განსხვავებული ფერით გამოყავით თითოეული კლასტერი

### Penguins Dataset

```python
# მონაცემების ჩატვირთვა
penguins = sns.load_dataset("penguins")
penguins = penguins.dropna()  

# Use numerical features for clustering
X = penguins[["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]]
```

### GMM
- გამოიყენეთ StandardScaler X სიმრავლის სკალირებისთვის
- ააგეთ BIC-გრაფიკი, კომპონენტების რაოდენობის შესარჩევად (n_components)
- სკალირებულ მონაცემებზე გაწვრთენით GaussianMixture ალგორითმი, გადაეცით შერჩეული n_components პარამეტრი
- ააგეთ გრაფიკი და განსხვავებული ფერით გამოყავით თითოეული კლასტერი
