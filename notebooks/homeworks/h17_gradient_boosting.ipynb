## **დავალებები: Gradient Boosting**

```python
import pandas as pd
```

#### Mushroom Dataset
Mushroom Dataset (UCI Machine Learning Repository-დან) შეიცავს ინფორმაციას სხვადასხვა ტიპის სოკოებზე, რომლებიც კლასიფიცირებულნი არიან როგორც საჭმელი (e) ან მომწამვლელი (p). მონაცემები შეიცავს 22 კატეგორიულ მახასიათებელს, როგორიცაა ქუდის ფორმა, სუნი და ფურცლების ფერი.

```python
# Load the dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/mushroom/agaricus-lepiota.data"
columns = [
    "class", "cap-shape", "cap-surface", "cap-color", "bruises", "odor",
    "gill-attachment", "gill-spacing", "gill-size", "gill-color", "stalk-shape",
    "stalk-root", "stalk-surface-above-ring", "stalk-surface-below-ring",
    "stalk-color-above-ring", "stalk-color-below-ring", "veil-type", "veil-color",
    "ring-number", "ring-type", "spore-print-color", "population", "habitat"
]
features = ["gill-color", "spore-print-color", "odor", "population", "gill-size", "stalk-root", 
 "stalk-surface-below-ring", "habitat", "ring-number", "stalk-color-below-ring", "class"
 ]
df_mushrooms = pd.read_csv(url, names=columns)[features]
df_mushrooms.head() 
```

სვეტი| მნიშვნელობა
--------------|------------------------
gill-color | ქუდის ქვედა lamella-ების ფერი (მაგ. თეთრი, ყავისფერი, შავი)
spore-print-color | სპორების ანაბეჭდის ფერი (ფერი, რომელიც რჩება ქაღალდზე როცა სოკო აწყვია)
odor | სუნი (მაგ. უსიამოვნო, ტკბილი, თევზის მსგავსი)
population | რაოდენობა ბუნებაში (მაგ. იშვიათი, ხშირი, ჯგუფური)
gill-size | lamella-ების ზომა (განიერი ან ვიწრო)
stalk-root | სოკოს ყუნწის ძირის ტიპი (მაგ. ძირიანი, ტალახში ჩაშვებული, არ არსებობს)
stalk-surface-below-ring | ყუნწის ზედაპირი ბეჭდის ქვემოთ (მაგ. გლუვი, ბოჭკოვანი, ქერცლიანი)
habitat | საცხოვრებელი გარემო (მაგ. ტყე, მდელო, ქალაქი, ბაღი)
ring-number | ბეჭდების რაოდენობა ყუნწზე (0, 1 ან 2 ბეჭედი)
stalk-color-below-ring | ყუნწის ფერი ბეჭდის ქვემოთ (მაგ. თეთრი, ყვითელი, მუქი)

### Gradient Boosting
  - გარდაქმენით კატეგორიული სვეტები ისე რომ მათი გამოყენება შეძლოთ მოდელის გაწვრთნისას
  - "class" არის სამიზნე სვეტი (y)
  - დაყავით მონაცემები training და testing სიმრავლეებად
  - გაწვრთენით Random Forest მოდელი training მონაცემებზე
      - გააკეთეთ პროგნოზი testing მონაცემებზე
      - მოახდინეთ მოდელის ევალუაცია
      - დაბეჭდეთ კლასიფიკაციის რეპორტი
      - ააგეთ confusion matrix
      - დაბეჭდეთ მახასიათებლების მნიშვნელობა (Feature importance)
  - გაწვრთენით Gradient Boosting მოდელი training მონაცემებზე
      - დააკვირდით ჰიპერპარამეტრებს `(n_estimators=100, learning_rate=0.1, max_depth=3)`
      - გააკეთეთ პროგნოზი testing მონაცემებზე
      - მოახდინეთ მოდელის ევალუაცია
      - დაბეჭდეთ კლასიფიკაციის რეპორტი
      - ააგეთ confusion matrix
      - დაბეჭდეთ მახასიათებლების მნიშვნელობა (Feature importance)
  - რომელი მოდელის პერფორმანსი იყო უკეთესი?
