## **დავალებები: PCA (Principal Component Analysis)¶**

```python
import pandas as pd
from sklearn.datasets import fetch_openml
```

ამ დავალებისთვის გამოვიყენებთ Leukemia Dataset მონაცემებს.

```python
# მონაცემების ჩატვირთვა
data = fetch_openml(name="leukemia", version=1, as_frame=True)
X = data.data
y = data.target

print(X.shape)  # (72 samples, 7129 gene features)
```

### PCA
- გამოიყენეთ StandardScaler მახასიათებლების გარდასაქმნელად
- შეამცირეთ მახასიათებლების რაოდენობა PCA-ს მეშვეობით
- აარჩიეთ კომპონენტების ოპტიმალური რაოდენობა
- ააგეთ პირველი ორი კომპონენტის scatterplot, მონაცემები გამოყავით კლასების შესაბამისი ფერით
- დაყავით მონაცემები training და testing სიმრავლეებად
- აარჩიეთ და გაწვრთენით მოდელი training მონაცემებზე
- გააკეთეთ პროგნოზი testing მონაცემებზე
- შეაფასეთ მოდელი, დაითვალეთ ევალუაციის მეტრიკები
- მოდელის ინტერპრეტაციისათვის გააკეთეთ უკუტრანსფორმაცია
- ააგეთ feature importance და shap value-ს გრაფიკები
