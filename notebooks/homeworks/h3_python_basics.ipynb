**1. ბოლო ციფრების ნამრავლი**

დაწერეთ ფუნქცია, რომელიც მიიღებს ორ მთელ რიცხვს და დააბრუნებს მათი ბოლო ციფრების ჯამს.

```python
def last_digit_sum(a, b):
    a = str(a)
    b = str(b)
    a_list = [int(num) for num in a]
    b_list = [int(num) for num in b]
    a_last = a_list[len(a)-1]
    b_last = b_list[len(b)-1]
    
    return a_last + b_last

print(last_digit_sum(123, 456))  # 3 + 6 = 9
print(last_digit_sum(789, 32))   # 9 + 2 = 11
```

**2. FizzBuzz ვარიაცია**

დაწერეთ ფუნქცია, რომელიც გადაყვება 1-დან გადმოცემულ რიცხვამდე და დაბეჭდავს:
- "Foo" - თუ რიცხვი იყოფა 4-ზე
- "Bar" - თუ რიცხვი იყოფა 6-ზე
- "FooBar" - თუ რიცხვი იყოფა 4-ზეც და 6-ზეც.

```python
def foo_bar(n):
    for num in range(1,n+1):
        if num %4 == 0:
            print(f"Foo - {num}")
        elif num %6 == 0:
            print(f"Bar - {num}")
        elif (num%4 ==0) & (num%6==0) :
            print(f"FooBar - {num}")
        else:
            print(f"Other - {num}")
            

foo_bar(20)
```

**3. შეცვლილი ფიბონაჩის მიმდევრობა**

დაწერეთ ფუნქცია, რომელიც დააბრუნებს შეცვლილი ფიბონაჩის მიმდევრობის მე-n წევრს, სადაც პირველი ორი წევრია 2 და 3, ხოლო ყოველი შემდეგი წევრი არის წინა ორი წევრის ჯამი.

```python

def modified_fibonacci(n):

    num_list = [2,3]
    for _ in range(0,n):
        new_num = sum(num_list[-2:])
        num_list.append(new_num)

    return num_list[n-1]


print(modified_fibonacci(5))  # 13



```

```python
import random

def guess_the_number():
    num_to_guess = random.randint(1,50)
    while 1 > 0 :
        user_num = int(input())    
        if user_num > num_to_guess:
            print(f"Go Lower (Less Than - {user_num})")
        elif user_num < num_to_guess:
            print(f"Go Higher (More Than {user_num})")
        else:
            break


guess_the_number()

```

**5. ციფრთა ნამრავლი**

დაწერეთ ფუნქცია, რომელიც მიიღებს 4-ნიშნა მთელ რიცხვს და დააბრუნებს მისი ციფრების ნამრავლს.

```python
def digit_product(n):
    num = str(n)
    result = 1

    for i in num:
        num_int = int(i)
        result = result * num_int
    return result

print(digit_product(1234))  # 1 * 2 * 3 * 4 = 24
```

**9. მოძრაობის სიმულაცია**

მომხმარებელი უნდა მართავდეს რიცხვს საწყისი წერტილიდან მიზნისკენ up და down ბრძანებებით (არა left/right).

```python
def move_to_goal(start, goal):
    while start != goal:
        if str(input()) == "up":
            start +=1
            print(f"Your Num is {start} And Goal Is {goal}")
        elif str(input()) == "Down":
            start -=1
            print(f"Your Num is {start} And Goal Is {goal}")

move_to_goal(0, 5)
```

**11. გამოთვლითი ამოცანა**
