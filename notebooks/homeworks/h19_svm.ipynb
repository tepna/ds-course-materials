### Support Vector Machines (SVMs)
- გამოიყენეთ StandardScaler მახასიათებლების გარდასაქმნელად
1. Heating Load

  - გაწვრთენით SVM მოდელი training მონაცემებზე
      - სცადე განსხვავებული kernel-ები: 'linear', 'poly', 'rbf'
      - გააკეთეთ პროგნოზი testing მონაცემებზე
      - მოახდინეთ მოდელის ევალუაცია: გამოითვალეთ ევალუაციის მეტრიკები
      - დაბეჭდეთ მახასიათებლების მნიშვნელობა (Feature importance)
      - შექმენით გრაფიკი, სადაც შეადარებთ test values vs. predicted values
