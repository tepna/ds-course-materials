## **დავალებები: Large Language Model (LLM)¶**

```python
import PyPDF2
```

```python
def extract_text_from_pdf(pdf_path):
    reader = PyPDF2.PdfReader(pdf_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text()
    return text
```
