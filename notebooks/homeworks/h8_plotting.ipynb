## **დავალებები: Plotting**

### Heart Disease Dataset
- **მონაცემების ფაილიდან წაკითხვა**
  - წაიკითხეთ მონაცემები heart_disease_uci.csv ფაილიდან (ფაილი მოთავსებულია: "../../datasets/heart_disease/heart_disease_uci.csv")..
  - გამოიყენეთ მეთოდები, როგორიცაა `.head()`, `.info()` და `.describe()` მონაცემთა ფორმატსა და შიგთავსზე საწყისი შთაბეჭდილების შესაქმნელად.
  - დააკვირდით, რომელი სვეტი წარმოადგენს სამიზნე ცვლადს (რომელი სვეტის პროგნოზს გააკეთებდით)
    
- **რიცხვითი მახასიათებლების ჰისტოგრამები**
  - გამოიყენეთ matplotlib ბიბლიოთეკა და შექმენით ჰისტოგრამები სხვადასხვა რიცხვითი მახასიათებლისთვის (მაგ., age, trestbps, chol).

- **scatter-plot ორ ციფრულ მახასიათებლებს შორის**
  - გამოიყენეთ matplotlib და ააგეთ scatter-plot, სადაც აჩვენებთ, როგორია დაკავშირებული ასაკი (age) და thalach.

- **bar-plot კატეგორიული მახასიათებლებისთვის**
  - შეამოწმეთ კატეგორიული მახასიათებლების (მაგ., target, sex, cp) განაწილება bar-plot გამოყენებით. გამოიყენეთ seaborn-ის countplot ფუნქცია.

- **კორელაციის heatmap**
  - შექმენით კორელაციის heatmap, რომელიც აჩვენებს ციფრული მახასიათებლების ურთიერთდამოკიდებულებას. გამოიყენეთ seaborn-ის heatmap ფუნქცია.
    
- **box-plot განწილების გამოსაკვლევად**
  - ააგეთ box-plot, რათა შეისწავლოთ სხვადასხვა მახასიათებლების განაწილება (მაგ. წონის განაწილება) დააკვირდით არის თუ არა დამოკიდებული target-ზე. გამოიყენეთ seaborn-ის boxplot.
 
- **Histogram** Plot a histogram of patients’ ages and color by heart disease (target).
    - გამოიყენეთ plotly და ააგეთ პაციენტების ასაკის განაწილების ჰისტოგრამა, გამოყავით "target" სვეტი სხვადასხვა ფერით (color="target"). რომელ ასაკობრივ ჯგუფშია გულის დაავადება უფრო ხშირი?
