## **დავალებები: Outlier detection¶**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_kddcup99, make_moons
```

### Synthetic Dataset

```python
# moons data (normal)
X_inliers, _ = make_moons(n_samples=300, noise=0.05, random_state=42)
X_outliers = np.random.uniform(low=-1.5, high=2.5, size=(30, 2)) # random outliers
X = np.vstack([X_inliers, X_outliers])
```

#### Local Outlier Factor (LOF)
- მონაცემებზე გაწვრთენით LOF ალგორითმი, სწორად შეარჩიეთ contamination პარამეტრი
- ააგეთ გრაფიკი, ფერით გამოყავით ნაპოვნი ანომალიები
- ააგეთ ჰისტოგრამა Distribution of LOF Scores
- ააგეთ LOF Score vs LRD და დარწმუნდით რომ თქვენს მიერ შერჩეული contamination პარამეტრი სწორია

### Network intrusion dataset

```python
# მონაცემების ჩატვირთვა
data = fetch_kddcup99(percent10=True, shuffle=True, as_frame=True)

df = data.frame
df.head()
```

```python
# features and target
features = [
    "duration","src_bytes","dst_bytes",
    "count","srv_count","same_srv_rate","diff_srv_rate",
    "dst_host_count","dst_host_srv_count"
]
X = df[features]
y = df["labels"]
y_binary = y.apply(lambda x: 1 if x == b"normal." else -1)
y_binary.value_counts()
```
