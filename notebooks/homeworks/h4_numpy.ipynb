## 8. 5x5 მატრიცა და სტანდარტიზაციის პროცესი

- **შექმენით 5x5 მატრიცა:**  
  შეავსეთ მატრიცა მთელი რიცხვებით (მაგ., `np.random.randint`-ის გამოყენებით).

- **სტანდარტიზაცია:**  
  გააკეთეთ სტანდარტიზაციის პროცესი, რაც ნიშნავს, რომ თითოეული ელემენტს გამოაკლდება მატრიცის საშუალო (mean) და ეს სხვაობა შეფარდდება მატრიცის სტანდარტულ გადახრასთან (standard deviation).  
  ფორმულით:

  
  $\text{standardized\_elem} = \frac{elem - \mu}{\sigma}$

  
  სადაც $\mu$ არის მატრიცის საშუალო და $\sigma$ – სტანდარტული გადახრა.

```python
# Your code here

rand_list = np.random.randint(1,10,size=[5,5])
rand_list_mean = rand_list.mean()
rand_list_std = rand_list.std()

rand_list_standartized = (rand_list - rand_list_mean)/rand_list_std
rand_list_standartized
```

## 9. ნორმალიზებული numpy array-ს დაბრუნება ფუნქციით

- დაწერეთ ფუნქცია, რომელსაც გადაეცემა numpy array და რომელიც აბრუნებს ნორმალიზირებულ მასივს.
  
- **ნორმალიზაციის პროცესი:**  
  თითოეული ელემენტი გარდაიქმნება შემდეგი ფორმულით:

  
$ \text{normalized\_elem} = \frac{elem - \min(arr)}{\max(arr) - \min(arr)} $

  სადაც $\min(arr)$ და $\max(arr)$ არის მოცემული მასივის მინიმალური და მაქსიმალური მნიშვნელობები.

```python
# Your code here

rand_list = np.random.randint(1,10,size=10)
rand_list_max = rand_list.max()
rand_list_min = rand_list.min()

nomrilized_elem  = (rand_list - rand_list_min) / (rand_list_max - rand_list_min)
nomrilized_elem
```
