## **დავალებები: Classification**

### 1. Logistic Regression

ამ დავალებისთვის გამოვიყენებთ ერთ-ერთი სარეკლამო კომპანიის მონაცემებს. 
აქ გვაქვს ინფორმაცია იმის შესახებ, გადავიდა თუ არა რეკლამის ბმულზე მომხმარებელი კომპანიის ვებ-გვერდიდან. 

მონაცემების ფაილი შეიცავს შემდეგ ინფორმაციას:
- Daily Time Spent on Site: მომხმარებლის მიერ ვებსაიტზე დახარჯული დღიური დრო წუთებში
- Age: მომხმარებლის ასაკი 
- Area Income: საშუალო შემოსავალი მომხმარებლის საცხოვრებელი რეგიონში
- Daily Internet Usage: მომხმარებლის ყოველდღიური ინტერნეტის გამოყენების საშუალო ხანგძლივობა
- Ad Topic Line: რეკლამის სათაური
- City: მომხმარებლის ქალაქი
- Male: აჩვენებს, მომხმარებელი არის თუ არა მამაკაცი
- Country: მომხმარებლის ქვეყანა
- Timestamp: დრო, როცა მომხმარებელი რეკლამის ბმულზე გადავიდა ან შესაბამისი რეკლამის ფანჯარა დახურა
- Clicked on Ad: 0 ან 1
