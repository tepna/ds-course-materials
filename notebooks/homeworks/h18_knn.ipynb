## **დავალებები: K-Nearest Neighbors (KNN)¶**

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

### 1.California Housing Data-თვის გამოიყენეთ SHAP Explainer მოდელის პერფორმანსის ასახსნელად. ააგეთ SHAP Summary Plot (plot_type="bar" პარამეტრით და მის გარეშე), ახსენით შედეგი.  

#### California Housing Data - სახლის ფასის პროგნოზი

```python
# მონაცემების ჩატვირთვა
data = fetch_california_housing()
X, y = data.data, data.target * 100000
feature_names = data.feature_names

df_housing = pd.DataFrame(X, columns=feature_names)
df_housing["MedHouseVal"] = y 
df_housing.head()
```

### 2. ამ დავალებისთვის გამოვიყენებთ Banknote Authentication Dataset მონაცემებს. კუპიურების სურათებიდან შექმნილია შემდეგი მახასიათებლები: variance, skewness, kurtosis, entropy.

```python
# ჩატვირთე მონაცემები
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt"
columns = ["variance", "skewness", "kurtosis", "entropy", "class"]
df = pd.read_csv(url, header=None, names=columns)
df.head()
```

ჩვენი მიზანია დავადგინოთ არის თუ არა კუპიურა ნამდვილი (ჩვენი მიზანია დავადგინოთ არის თუ არა კუპიურა ნამდვილი (ნამდვილი (0), ყალბი (1)).
