```python
import os
import pandas as pd
import pymysql # for engineecting to and interacting with a MySQL database 
from dotenv import load_dotenv #.env ფაილის ჩასატვირთად 
from sqlalchemy import create_engine # მართავს კავშირს მონაცემთა ბაზასთან
```

```python
# Keep secrets out of code
# ჩატვირთე .env ფაილი 
load_dotenv()

# წაიკითხე ცვლადები გარემოდან
db_config = {
    "host": os.getenv("DB_HOST"),
    "user": os.getenv("DB_USER"),
    "database": os.getenv("DB_NAME")
}
```

```python
# Generic connection helper
# SQLAlchemy engine შექმნა (engineection pool Python-ისა და მონაცემთა ბაზას შორის)
engine = create_engine(
    f"mysql+pymysql://{db_config["user"]}@{db_config["host"]}/{db_config["database"]}"
)
```

```python
query = "SHOW FULL TABLES;"
tables_df = pd.read_sql(query, engine)
tables_df.head()
```

```python
# Sakila ცხრილების წაკითხვა pandas-ს გამოყენებით
actor_df = pd.read_sql("SELECT * FROM actor;", engine)
film_df = pd.read_sql("SELECT * FROM film;", engine)
customer_df = pd.read_sql("SELECT * FROM customer;", engine)
rental_df = pd.read_sql("SELECT * FROM rental;", engine)
payment_df = pd.read_sql("SELECT * FROM payment;", engine)
```

```python
# SQL Select Top: პირველი 10 ჩანაწერის ჩვენება
actor_df.head(10)
```

```python
# SQL Count: ჩანაწერების რაოდენობა
print(f"Total number of actors: {actor_df.shape[0]}")
```

```python
# SQL Count: ჩანაწერების რაოდენობა
total_films = film_df.shape[0] 
print(f"Total number of films: {total_films}")
```

```python
null_address_df = actor_df[actor_df['last_name'].isnull()]
null_address_df
```

```python
# SQL Select: კონკრეტული სვეტების ამოღება
selected_columns = actor_df[["first_name", "last_name"]]
selected_columns.head()
```

```python
# SQL Select Distinct: უნიკალური მნიშვნელობების მიღება
distinct_last_names = actor_df["last_name"].unique()
distinct_last_names
```

```python
# SQL Where: ფილტრაცია
filtered_films = film_df[film_df["rating"] == "PG-13"]
filtered_films.head(3)
```

```python
# SQL Order By: დალაგება
sorted_customers = customer_df.sort_values(by="last_name", ascending=True)
sorted_customers.head()
```

```python
# SQL Order By: დალაგება
sorted_customers = customer_df.sort_values(by="last_name", ascending=False)
sorted_customers.head()
```

```python
# SQL And: რამდენიმე პირობა
filtered_customers = customer_df[(customer_df["active"] == 1) & (customer_df["store_id"] == 1)]
filtered_customers.head()
```

```python
# SQL Or: ერთი პირობის შესრულება საკმარისია
filtered_films_or = film_df[(film_df["rating"] == "R") | (film_df["rental_duration"] > 5)]
filtered_films_or.head(3)
```

```python
# SQL Not: პირობის უარყოფა
filtered_customers_not = customer_df[~(customer_df["active"] == 1)]
filtered_customers_not.head()
```

```python
# SQL Min and Max
min_rental = film_df["rental_rate"].min()
max_rental = film_df["rental_rate"].max()

print(f"Minimum rental rate: {min_rental}")
print(f"Maximum rental rate: {max_rental}")
```

```python
# SQL Max
max_rental_duration = film_df.loc[film_df["rating"] == "R", "rental_duration"].max()
print(max_rental_duration)
```

```python
# SQL Sum: მონაცემების ჯამი
print(f"Total payment amount: {payment_df["amount"].sum()}")
```

```python
# SQL Avg: საშუალო მნიშვნელობა
print(f"Average rental duration: {film_df["rental_duration"].mean()}")
```

```python
# SQL In: ფილტრაცია
film_df[film_df["rating"].isin(["PG", "G"])].head(3)
```

```python
# SQL Between: მნიშვნელობათა დიაპაზონი
payment_df[payment_df["amount"].between(5,10)]
```

```python
# SQL Group By: მონაცემთა დაჯგუფება
grouped_ratings = film_df.groupby("rating").size().reset_index(name="count")
grouped_ratings.head()
```

```python
# SQL Having: ფილტრაცია დაჯგუფებული მონაცემებისთვის
grouped_ratings = film_df.groupby("rating").size().reset_index(name="count")
filtered_grouped_ratings = grouped_ratings[grouped_ratings["count"] > 200]
filtered_grouped_ratings
```

```python
# SQL Exists: არსებული მონაცემის შემოწმება
customer_exists = customer_df[customer_df["customer_id"].isin(rental_df["customer_id"])]
customer_exists.head()
```

```python
# SQL Joins: კავშირი ორ DataFrame-ს შორის
customer_rentals = customer_df.merge(rental_df, on="customer_id")[["first_name", "rental_date"]]
customer_rentals.head()
```

```python
# SQL Joins: კავშირი ორ DataFrame-ს შორის
customer_rentals = customer_df.merge(rental_df, on="customer_id", how="inner")[["first_name", "rental_date"]]
customer_rentals.head()
```

```python
# SQL Left Join
customer_rentals_left = customer_df.merge(rental_df, on="customer_id", how="left")[["first_name", "rental_date"]]
customer_rentals_left.head()
```

```python
# SQL Right Join
customer_rentals_right = rental_df.merge(customer_df, on="customer_id", how="right")[["first_name", "rental_date"]]
customer_rentals_right.head()
```

```python
# SQL Full Join: ყველა ჩანაწერი ორივე ცხრილიდან
customer_rentals_full = rental_df.merge(customer_df, on="customer_id", how="outer")[["first_name", "rental_date"]]
customer_rentals_full.head()
```

```python
# SQL Union: ორი DataFrame-ის გაერთიანება
union_df = pd.concat([actor_df[["first_name"]], customer_df[["first_name"]]]).drop_duplicates()
union_df.head()
```

```python
# SQL Case: პირობის მიხედვით მნიშვნელობის არჩევა
film_df["price_category"] = film_df["rental_rate"].apply(lambda x: "Cheap" if x < 2 else ("Moderate" if 2 <= x <= 4 else "Expensive"))
film_df.head(3)
```

```python
# SQL Insert Into: მონაცემების ჩასმა
new_row = {"first_name": "John", "last_name": "Doe"}
actor_df = pd.concat([actor_df, pd.DataFrame([new_row])], ignore_index=True)
actor_df.tail()
```

```python
actor_df.loc[actor_df["first_name"] == "John", "last_name"] = "Smith"
actor_df[actor_df["first_name"] == "John"]
```

```python
actor_df = actor_df[~((actor_df["first_name"] == "John") & (actor_df["last_name"] == "Smith"))]
actor_df[((actor_df["first_name"] == "John") & (actor_df["last_name"] == "Smith"))]
```

```python
max_g_duration = film_df.loc[film_df["rating"] == "G", "rental_duration"].max()
film_df[film_df["rental_duration"] > max_g_duration].head()
```
