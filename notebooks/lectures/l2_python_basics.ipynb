#### **ოპერაციები ცვლადებზე**

არითმეტიკული ოპერაციები:

```python
a = 12
b = 5

# შეკრება
sum_result = a + b  # 17

# გამოკლება
sub_result = a - b  # 7

# გამრავლება
mul_result = a * b  # 60

# გაყოფა
div_result = a / b  # 2.4

# მთელ რიცხვზე გაყოფა (განაყოფის მთელი ნაწილი)
floor_div_result = a // b  # 2

# ნაშთი გაყოფის შემდეგ
mod_result = a % b  # 2

# ხარისხში აყვანა
exp_result = a ** b  # 248832
```

```python
# Printing the results
print("Sum:", sum_result)
print("Subtraction:", sub_result)
print("Multiplication:", mul_result)
print("Division:", div_result)
print("Floor Division:", floor_div_result)
print("Modulus:", mod_result)
print("Exponentiation:", exp_result)
```

*   ელემენტების დამატება ხდება append(), insert() მეთოდებით
*   ელემენტების წაშლა ხდება pop(), remove() მეთოდებით

```python
my_list = [1, 2, 3, "apple", 4.5]

# ელემენტის დამატება
my_list.append(6)
my_list.insert(2, "banana")  # მეორე პოზიციაზე ახალი ელემენტი

# ელემენტის წაშლა
my_list.pop()  # ბოლო ელემენტს წაშლის
my_list.remove("apple")  # "apple" წაშლის

print(my_list)  

# ინდექსირება და სლაისინგი
print(my_list[0])  # პირველი ელემენტი: 1
print(my_list[-1])  # უკანასკნელი ელემენტი: 4.5
print(my_list[1:3])  # 2, "banana"
```

```python
names = ["Tim", "Elon", "Jenny", "Bill", "Melanie", "Melissa"]  # სტრინგების სია
ages = [19, 37, 40, 62, 31, 33]  # მთელი რიცხვების სია

print(sorted(ages)) # დალაგება
print(sum(ages))  # ჯამი
print(sorted(names))  # ალფაბეტური დალაგება
print(", ".join(names))  # მიერთება, ყველა ელემენტი ერთ სტრინგში, გამყოფი ღირებულებით: ", "
avg_age = sum(ages) / len(ages)  # საშუალო ასაკი
print(f"The average age is {avg_age} years.")
```

#### Set
Set არის მონაცემთა სტრუქტურა, რომელიც ინახავს უნიკალურ ელემენტებს (duplicated ელემენტები არ შეიძლება არსებობდეს). ის არის mutable (შეცვლადი).

```python
my_set = {1, 2, 3, 4, 25}

# ელემენტის დამატება
my_set.add(6)

# ელემენტის წაშლა
my_set.remove(3)

print(my_set) # შედეგი: {1, 2, 4, 5, 6}
```

```python
my_set = {1, 2, 3, 4, 5}

# მათემატიკური ოპერაციები
another_set = {4, 5, 6, 7}
union_set = my_set.union(another_set)  # შეუერთდება (ყველა ელემენტი)
intersection_set = my_set.intersection(another_set)  # კვეთა (საერთო ნაწილი)
difference_set = my_set.difference(another_set)  # სხვაობა (ის ელემენტები რაც მხოლოდ პირველ სიმრავლეშია)

print(union_set)  # შედეგი: {1, 2, 4, 5, 6, 7}
print(intersection_set)  # შედეგი: {4, 5}
print(difference_set) # შედეგი: {1, 2, 3}
```

#### Dictionary
Dictionary-ი არის მონაცემთა სტრუქტურა, რომელიც შეიცავს key-value წყვილებს. key არის უნიკალური, ხოლო value შეიძლება იყოს ნებისმიერი ტიპი. ის არის mutable (შეცვლადი) და გამოიყენება კონკრეტული ელემენტების სწრაფად მისაღებად.

*   enumerate(): ეხმარება ინდექსის (პოზიციის) დამატებაში.
*   zip(): საშუალებას იძლევა, რომ ერთად გაიაროთ ორი ან მეტი სია და ამ სიის შესაბამისი ელემენტები დააწყვილოთ.

```python
names = ["Tim", "Elon", "Jenny", "Bill", "Melanie", "Melissa"]
ages = [19, 37, 40, 62, 31, 33]

# მხოლოდ სახელების გადავლა
for name in names:
    print(name)

# enumerate გამოიყენება ინდექსის დამატებისთვის
for idx, name in enumerate(names):
    print(f"{name} at position {idx}.")

# zip გამოიყენება ორი სიის ერთდროულად გასავლელად
for age, name in zip(ages, names):
    print(f"{name} is {age} years old.")
```

### წრფივი ალგებრა (ვექტორები & მატრიცები)

#### 1. ვექტორები

ვექტორი არის რიცხვების დალაგებული სია (აქვს მიმართულება და სიგრძე).

$v = \begin{bmatrix} 1 \ 5 \end{bmatrix}$ არის 2D ვექტორი → მარჯვნივ 1 ერთეული და ზემოთ 5 ერთეული.

<img src="../../images/vectors.png" alt="numpy" width=400>

მონაცემთა მეცნიერებაში, ვექტორი შეიძლება იყოს მონაცემების ერთი სტრიქონი(მაგ., `[სიმაღლე, წონა, ასაკი]`).

---

2 განზომილებიანი ვექტორებისთვის (2D vectors) მნიშვნელოვანი ფორმულები ($a = (a_1, a_2), \quad b = (b_1, b_2)$):

შეკრება (Addition):

$a + b = (a_1 + b_1,  a_2 + b_2)$


გამოკლება (Subtraction):

$a - b = (a_1 - b_1,  a_2 - b_2)$

<img src="../../images/vec.png" alt="My Image" width="400"/>

---

სკალირება (Scaling by scalar (k)):

$k \cdot a = (k \cdot a_1, k \cdot a_2)$

---

წერტილოვანი ნამრავლი (Dot Product) → გამოიყენება მსგავსების გაზომვაში:

$a \cdot b = a_1 b_1 + a_2 b_2$

---

ჯვარედინი ნამრავლი (Cross Product in 2D → scalar) → აბრუნებს პარალელოგრამის ფართობს, რომელსაც ქმნიან (a) და (b):
$a \times b = a_1 b_2 - a_2 b_1$

---

ნორმა (Norm, ვექტორის სიგრძე):

$|a| = \sqrt{a_1^2 + a_2^2}$

---

#### 2. მატრიცები

მატრიცა არის რიცხვების მართკუთხა ცხრილი.

$$A =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}$$

მატრიცების შეკრება (Addition):

შეიძლება მხოლოდ მაშინ, თუ ორივე მატრიცას აქვს იგივე განზომილება (იგივე რიგების და სვეტების რაოდენობა).

$$
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
+
\begin{bmatrix}
5 & 6 \\
7 & 8
\end{bmatrix}
=
\begin{bmatrix}
6 & 8 \\
10 & 12
\end{bmatrix}
$$

---

მატრიცების ნამრავლი (Multiplication):

შეიძლება მხოლოდ მაშინ, თუ პირველი მატრიცის სვეტების რაოდენობა ტოლია მეორე მატრიცის რიგების რაოდენობის.

თუ $A$ არის $m \times n$ და $B$ არის $n \times p$, მაშინ ნამრავლი $A \cdot B$ იქნება $m \times p$ მატრიცა.

$$
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}_{2 \times 2}
\cdot
\begin{bmatrix}
5 & 6 \\
7 & 8
\end{bmatrix}_{2 \times 2}
=
\begin{bmatrix}
19 & 22 \\
43 & 50
\end{bmatrix}_{2 \times 2}
$$

$$
\begin{bmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{bmatrix}_{2 \times 3}
\cdot
\begin{bmatrix}
7 \\
8 \\
9
\end{bmatrix}_{3 \times 1}=
\begin{bmatrix}
1\cdot 7 + 2\cdot 8 + 3\cdot 9 \\
4\cdot 7 + 5\cdot 8 + 6\cdot 9
\end{bmatrix}_{2 \times 1}=
\begin{bmatrix}
50 \\
122
\end{bmatrix}_{2 \times 1}
$$

---

ტრანსპონირება (Transpose):

სტრიქონები და სვეტები ერთმანეთს ეცვლება.  
თუ $A$ არის $m \times n$, მაშინ $A^T$ იქნება $n \times m$.

$$
A^T =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}^T
=
\begin{bmatrix}
1 & 3 \\
2 & 4
\end{bmatrix}
$$

---

ინვერსია (Inverse):

$A^{-1}$ არსებობს მხოლოდ კვადრატული მატრიცებისთვის ($n \times n$), თუ $det(A) \neq 0$.  
შესრულებულია: $A \cdot A^{-1} = I$

$$
A^{-1} = \frac{1}{ad - bc}
\begin{bmatrix}
d & -b \\
-c & a
\end{bmatrix},
\quad თუ \quad  A = 
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}.
$$
