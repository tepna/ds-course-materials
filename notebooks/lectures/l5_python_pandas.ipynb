### DataFrame

Pandas DataFrame არის მონაცემთა სტრუქტურა, რომელიც გამოიყენება Python-ში მონაცემების ანალიზისთვის.

```python
data = {
    "ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35]
}

# DataFrame-ის შექმნა dictionary-დან
df = pd.DataFrame(data)
df
```

### DataFrame-ების გაერთიანება

```python
# merge() გამოიყენება, როცა გვინდა ორი DataFrame გაერთიანდეს SQL-ის JOIN ოპერაციის მსგავსად
    
# პირველი DataFrame
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

# მეორე DataFrame
df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Age': [25, 30, 35]
})

# DataFrame-ების გაერთიანება 'ID' სვეტზე (inner join)
merged_df = pd.merge(df1, df2, on='ID', how='inner')
merged_df
```

```python
# left join
merged_df = pd.merge(df1, df2, on='ID', how='left')
merged_df
```

```python
# right join
merged_df = pd.merge(df1, df2, on='ID', how='right')
merged_df
```

```python
# outer join
merged_outer_df = pd.merge(df1, df2, on='ID', how='outer')
merged_outer_df
```

```python
df1.merge(df2, on='ID', how='outer')
```

```python
# concat() საშუალებას გვაძლევს ერთობლივად შევინახოთ DataFrame-ები

# პირველი DataFrame
df1 = pd.DataFrame({
    'A': [1, 2],
    'B': [3, 4]
})

# მეორე DataFrame
df2 = pd.DataFrame({
    'C': [5, 6],
    'D': [7, 8]
})

# DataFrame-ების გაერთიანება სვეტების მიხედვით
concat_df = pd.concat([df1, df2], axis=1)
concat_df
```

```python
# DataFrame-ების გაერთიანება რიგების მიხედვით
concat_rows_df = pd.concat([df1, df2], axis=0)
concat_rows_df
```

```python
# join() ძირითადად გამოიყენება ინდექსის მიხედვით გაერთიანებისთვის

# პირველი DataFrame
df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
}, index=['a', 'b', 'c'])

# მეორე DataFrame
df2 = pd.DataFrame({
    'City': ['New York', 'Los Angeles', 'Chicago']
}, index=['a', 'b', 'c'])

# DataFrame-ების გაერთიანება ინდექსის მიხედვით
joined_df = df1.join(df2)
joined_df
```
